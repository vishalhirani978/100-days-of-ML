```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as mlt 
```

```python
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier

from sklearn.metrics import accuracy_score 
from sklearn.model_selection import cross_val_score

from sklearn.preprocessing import KBinsDiscretizer
from sklearn.compose import ColumnTransformer
```

```python
df = pd.read_csv('train(ML).csv',usecols=['Age','Fare','Survived'])
```

```python
df.dropna(inplace=True)
```

```python

df.head()
```

```python
x =df.iloc[:,1:3]
y= df.iloc[:,0]
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
clf =DecisionTreeClassifier()
```

```python
clf.fit(x_train,y_train)
y_pred =clf.predict(x_test)
```

```python
accuracy_score(y_test,y_pred)
```

```python
np.mean(cross_val_score(DecisionTreeClassifier(),x,y,cv=10,scoring='accuracy'))
```

```python
#now lets check with discretizer
kbins_age = KBinsDiscretizer(n_bins=5,encode='ordinal',strategy='quantile') 
bbins_fare = KBinsDiscretizer(n_bins=5,encode='ordinal',strategy='quantile')
```

```python
trf =ColumnTransformer([
    ('first',kbins_age,[0]),
    ('second',bbins_fare,[1])
]) 
```

```python
x_trans= trf.fit_transform(x_train)
x_test_trf = trf.transform(x_test)
```

```python
trf.named_transformers_['second'].n_bins_
```

```python
trf.named_transformers_['second'].bin_edges_
```

```python
output = pd.DataFrame({
    'age':x_train['Age'],
    'age_trf':x_trans[:,0],
    'fare':x_train['Fare'],
    'fare_trf':x_trans[:,1]
})
```

```python
output['age_labels'] = pd.cut(x=x_train['Age'],
                            bins=trf.named_transformers_['first'].bin_edges_[0].tolist())
output['fare_labels'] = pd.cut(x=x_train['Fare'],
                            bins=trf.named_transformers_['second'].bin_edges_[0].tolist())
```

```python
output.sample(5)
```

```python
clf =DecisionTreeClassifier()
clf.fit(x_trans,y_train)
y_pred2 =clf.predict(x_test_trf)
```

```python
accuracy_score(y_test,y_pred2)
```
