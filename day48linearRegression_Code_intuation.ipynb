```python
import  pandas as pd 
import matplotlib.pyplot as mlt 
import numpy as np
```

```python
df = pd.read_csv('placement.csv')
```

```python
df.head()
```

```python
mlt.scatter(df['cgpa'],df['package'])
mlt.xlabel('cgpa')
mlt.ylabel('package')
```

```python
X = df.iloc[:,0:1]
Y = df.iloc[:,:-1]
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,Y_train,Y_test = train_test_split(X,Y,test_size=0.2,random_state=2)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
lr =LinearRegression()
```

```python
lr.fit(X_train,Y_train)
```

```python
X_test
```

```python
Y_test
```

```python
lr.predict(X_test.iloc[2].values.reshape(1,1))
```

```python
mlt.scatter(df['cgpa'],df['package'])
mlt.plot(X_train,lr.predict(X_train),color='red')
mlt.xlabel('CGPA')
mlt.ylabel('PAckage')
```
