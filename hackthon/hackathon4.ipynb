```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

# Our clean dataset from Day 3
data = {
    "Name": ["Ali", "Sara", "Zara", "Ahmed"],
    "Age": [22, 25, 67.5, 23],
    "Score": [88, 92, 95, 82.5],
    "City": ["Karachi", "Lahore", "Islamabad", "Lahore"],
    "Grade": ["B", "A", "A", "B"]
}

df = pd.DataFrame(data)
print("Dataset ready!")
print(df)
```

```python
plt.figure(figsize=(8, 5))
plt.bar(df["Name"], df["Score"], color=["blue", "green", "red", "orange"])
plt.title("Student Scores Comparison")
plt.xlabel("Student Name")
plt.ylabel("Score")
plt.ylim(70, 100)
plt.grid(axis="y", linestyle="--", alpha=0.7)
plt.show()
```

```python
# Pie Chart - Students per City
city_counts = df["City"].value_counts()

plt.figure(figsize=(6, 6))
plt.pie(city_counts, 
        labels=city_counts.index, 
        autopct="%1.1f%%",
        colors=["skyblue", "lightgreen", "orange"],
        startangle=90)
plt.title("Students by City")
plt.show()
```

```python
# Line Chart - Age vs Score
plt.figure(figsize=(8, 5))
plt.plot(df["Name"], df["Score"], 
         marker="o", 
         color="purple", 
         linewidth=2,
         markersize=8)
plt.title("Score Trend by Student")
plt.xlabel("Student")
plt.ylabel("Score")
plt.grid(True, linestyle="--", alpha=0.7)
for i, score in enumerate(df["Score"]):
    plt.text(i, score + 0.5, str(score), ha="center")
plt.show()
```

```python
plt.figure()
colors = ['red', 'blue', 'green', 'orange']
plt.bar(df['Name'], df['Age'], color=colors)
#Title and labels
plt.title("Student Ages")
plt.xlabel("Students")
plt.ylabel("Age")
```
