```python
import pandas as pd
import numpy as np

# MESSY dataset like real world data!
data = {
    "Name": ["Ali", "Sara", None, "Zara", "Ahmed", "Sara", "Bilal"],
    "Age": [22, 25, 23, None, 23, 25, 200],
    "Score": [88, 92, 75, 95, None, 92, 55],
    "City": ["karachi", "Lahore", "KARACHI", "Islamabad", "lahore", "Lahore", None]
}

df = pd.DataFrame(data)
print("Our Messy Dataset:")
print(df)
print("\nShape:", df.shape)
```

```python
# Find missing values
print("Missing values in each column:")
print(df.isnull().sum())

print("\nTotal missing values:", df.isnull().sum().sum())

# Find duplicates
print("\nNumber of duplicate rows:", df.duplicated().sum())
print("\nThe duplicate rows:")
print(df[df.duplicated()])
```

```python
# Fix 1 - Remove duplicates
df = df.drop_duplicates()
print("After removing duplicates:")
print(df)
print("New shape:", df.shape)
```

```python
# Fix 2 - Drop rows with missing Name
df = df.dropna(subset=["Name"])
print("After dropping missing Name:")
print(df)
```

```python
# Fix 3 & 4 - Fill missing Age and Score with average
df["Age"] = df["Age"].fillna(df["Age"].mean())
df["Score"] = df["Score"].fillna(df["Score"].mean())
print("After filling missing Age and Score:")
print(df)
```

```python
# Fix 5 - Remove impossible age
df = df[df["Age"] <= 100]

# Fix 6 - Standardize city names
df["City"] = df["City"].str.title()
df["City"] = df["City"].fillna("Unknown")

print("✅ FINAL CLEAN DATASET:")
print(df)
print("\nMissing values remaining:", df.isnull().sum().sum())
print("Shape:", df.shape)
```

```python
df["Grade"] = df["Score"].apply(lambda x: "A" if x >= 90 else ("B" if x >= 75 else "C"))
print(df)
```
