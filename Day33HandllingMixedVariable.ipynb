```python
import pandas as pd 
import numpy as np 
```

```python
df =pd.read_csv('titanic.csv')
```

```python
df['number'].unique()
```

```python
fig =df['number'].value_counts().plot.bar()
fig.set_title('passenger travelling with')
```

```python
# when we  use feature that have both categorical numrical values so we will do sepration 
#extracting numrical part 
df['number_numrical'] = pd.to_numeric(df['number'],errors='coerce',downcast='integer')
```

```python
#now extracting categoricalsection 
df['number_categorical'] =np.where(df['number_numrical'].isnull(),df['number'],np.nan)
df.head(6)
```

```python
# ok now how we will seprate for cabin and ticket 
df['Cabin'].unique()
```

```python
df['cabin_num'] = df['Cabin'].str.extract('(\d+)')# captures numrical part 
df['cabin_cat'] = df['Cabin'].str[0]#capture the first letter 
df.head(6)
```

```python
df['cabin_cat'].value_counts().plot(kind='bar')
```

```python
#smilarly  now for ticket 
#extract the last bit of ticcket  as number 
df ['ticket_num'] = df['Ticket'].apply(lambda s:s.split()[-1])
df['ticket_num'] = pd.to_numeric(df['ticket_num'],errors='coerce',downcast='integer')

#extract the first part of ticket as category 
df['ticket_cat'] = df['ticket'].apply(lambda s:s.split()[0])
df['ticket_cat'] = np.where(df['ticket'].str.isdigit(),np.nan,df['ticket_cat'])
df.head()
```
