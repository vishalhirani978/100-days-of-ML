```python
def add_all(*args):
    return sum(args)

print(add_all(2, 3, 4, 5))
```

## Exercise 5

Write a function called `show_student` that accepts any keyword arguments using `**kwargs` and prints each key and value.
Try it with a name, age, and city.

```python
def show_student(**kwargs):
    for key, value in kwargs.items():
        print(f'{key}: {value}')

show_student(name='Ali', age=20, city='Lahore')
```

### Exercise 6: Feature Normalization

Write a function `min_max_scale(values)` that scales a list of numbers to range 0 to 1.
Use this formula:

scaled = (x - min_value) / (max_value - min_value)

If all values are the same, return a list of zeros.

```python
def min_max_scale(values):
    minimum = min(values)
    maximum = max(values)

    if minimum == maximum:
        return [0 for _ in values]

    return [(x - minimum) / (maximum - minimum) for x in values]

sample_feature = [120, 150, 180, 210]
print(min_max_scale(sample_feature))
```

### Exercise 7: Split Features and Labels

Given rows like `[hours_studied, attendance, result]`, write a function `split_features_labels(rows)`.
It should return two lists:
- features: each row except the last value
- labels: the last value from each row

```python
def split_features_labels(rows):
    features = []
    labels = []

    for row in rows:
        features.append(row[:-1])
        labels.append(row[-1])

    return features, labels

dataset = [
    [2, 80, 0],
    [5, 90, 1],
    [1, 70, 0],
    [6, 95, 1]
]

X, y = split_features_labels(dataset)
print('Features:', X)
print('Labels:', y)
```

### Exercise 8: Accuracy Score

Write a function `accuracy_score(y_true, y_pred)` that returns model accuracy in percent.

Formula:

accuracy = correct_predictions / total_predictions * 100

```python
def accuracy_score(y_true, y_pred):
    correct = 0

    for actual, predicted in zip(y_true, y_pred):
        if actual == predicted:
            correct += 1

    return (correct / len(y_true)) * 100

actual = [1, 0, 1, 1, 0]
predicted = [1, 0, 0, 1, 0]
print('Accuracy:', accuracy_score(actual, predicted), '%')
```
