<h2>WORKING WITH DATES<h2/>

```python
# fisrtly we will focus on date dataset
#converting into date time data type 
date['date']=pd.to_datetime(date['date'])
```

```python
date.info()
```

<h3>EXRACT YEAR</h3>

```python
date['date_year']=date['date'].dt.year
date.head(6)
```

<h3>EXTRACT MONTH<h3/>

```python
date['date_month'] =date['date'].dt.month 
date.head(6)
```

```python
date['date_month'] =date['date'].dt.month_name()
date.head(6)
```

```python
date['date_dayname'] =date['date'].dt.day_name()
date.head(6)
```

```python
date['date_day_week'] =date['date'].dt.dayofweek
date.head(6)
```

```python
# is weekend ..?
date['date_is_weakend'] =np.where(date['date_dayname'].isin(['Sunday','Saturday']),1,0)
date.drop(columns=['product_id','city_id','orders']).head(6)
```

<h3>EXTRACT WEAK OF THE YEAR<h3/>

```python
date['date_week']=date['date'].dt.week
```

<h3>EXTRACT SEMSTER<h3/>

```python
date['senster']=np.where(date['quarter'].isin([1,2]),1,2)
date.drop(columns=['product_id','city_id','orders']).head(6)
```
