<h2>let's start our own LR (for multiple regression)</h2>

```python
class MeraLR:
    def __init__(self):
        self.coef_ =None 
        self.intercept_ =None 
    def fit (self,X_train,y_train):
        X_train = np.insert(X_train,0,1,axis=1)
        # now will calculate the coeff
        betas = np.linalg.inv(np.dot(X_train.T,X_train)).dot(X_train.T).dot(y_train)
        self.intercept_ = betas[0]
        self.coef_ = betas[1:]
    def predict(self,X_test):
        y_pred = np.dot(X_test,self.coef_) + self.intercept_
        return y_pred
```

```python
lr = MeraLR()
```

```python
lr.fit(x_train,y_train)# here we have everything  but we just need to add '1' column we will use np.insert() 
```

```python
y_pred =lr.predict(x_test)
```

```python
r2_score(y_test,y_pred)
```
