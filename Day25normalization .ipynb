```python
#most of time min max is considerd as  normalization but actually it is tyoe of normalization  other types are [mean normalization, max absolute normalization,  robust scalling and many more ]
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as  sns 
```

```python
df = pd.read_csv('wine_data.csv',header =None ,usecols = [0,1,2])
df.columns =['Class label','Alchol','Malic acid']
```

```python
sns.kdeplot(df['Alchol'])
```

```python
sns.kdeplot(df['Malic acid'])
```

```python
sns.scatterplot(x=df['Alchol'],y= df['Malic acid'],hue=df['Class label'] )
```

<h2>train test split <h2/>

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test =train_test_split(df.drop(['Class label'],axis=1),df['Class label'],test_size = 0.3 ,random_state = 0)
X_train.shape,X_test.shape
```

```python
from sklearn.preprocessing import  MinMaxScaler  
scaler = MinMaxScaler()
#fit the scaler to the train set ,it will learn parametre 
scaler.fit(X_train)
#transform tarin and test sets 
X_train_scaled = scaler.transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
X_train_scaled = pd.DataFrame(X_train_scaled,columns = X_train.columns)
X_test_scaled =pd.DataFrame(X_test_scaled,columns =X_test.columns )
```

```python
np.round(X_train.describe(),1)
```

```python
np.round(X_train_scaled.describe(),1)
```

```python
fig,(ax1,ax2)=plt.subplots(ncols=2 ,figsize =(12,5))
ax1.set_title ("before scaling ")
ax2.set_title('after scalling ')
ax1.scatter(X_train['Alchol'], X_train['Malic acid'] ,c=y_train)
ax2.scatter(X_train_scaled['Alchol'], X_train_scaled['Malic acid'] ,c=y_train)
```

```python
fig,(ax1,ax2)=plt.subplots(ncols=2 ,figsize =(12,5))
ax1.set_title ("before scaling ")
ax2.set_title("Fter scalling ")
sns.kdeplot(X_train['Alchol'],ax =ax1)
sns.kdeplot(X_train['Malic acid'],ax=ax1)

sns.kdeplot(X_train_scaled['Alchol'],ax=ax2)
sns.kdeplot(X_train_scaled['Malic acid'],ax=ax2)
```
